### 1. 기본 수학 함수

| 함수 | 의미 | 예시 | 설명 |
| :---: | :---: | :---: | :---: |
| torch.abs(a) | 절대값 | abx(-3) -> 3 | 모든 원소의 절대값 반환 |
| torch.sqrt(a) | 제곱근 | sqrt(9) -> 3 | $\sqrt(x)$ |
| torch.square(a) | 제곱 | square(3) -> 9| $x^2$ |
| torch.pow(a, n) | 거듭제곱 | pow(2,3) -> 8 | $x^n$ |
| torch.exp(a) | 지수함수 | exp(1) -> e | $e^x$ |
| torch.log(a) | 자연로그 | log(e) -> 1 | ln(x) |
| torch.log10(a) | 상용로그 | log10(10) -> 1 | $log_{10}(x)$ |
| torch.log2(a) | 밑이 2인로그 | log2(2) -> 1 | $log_{2}(x)$ |

```python
a = torch.randn(3,3)
print(a)

# 기본 수학함수
print(torch.abs(a))
print(torch.sqrt(torch.abs(a)))
print(torch.square(a))
print(torch.pow(a, 2))
print()
# 지수함수
print(torch.exp(a))
print(torch.log(torch.abs(a))) # logₑa
print(torch.log(torch.exp(torch.tensor(1))))
print(torch.log(torch.exp(torch.tensor(a)))) # logₑeᵃ
print(torch.log10(a))
print(torch.log2(a))
```

### 3. 최솟값 / 최댓값

| 함수 | 의미 | 설명 |
| :---: | :---: | :---: |
| torch.min(a) | 최솟값 | 전체원소 중 가장 작은 값 |
| torch.max(a) | 최댓값 | 전체원소 중 가장 큰 값 |
| torch.argmin(a) | 최솟값의 인덱스 | 최솟값이 위치한 인덱스 반환 |
| torch.argmax(a) | 최댓값의 인덱스 | 최댓값이 위치한 인덱스 반환 |

```python
a = torch.randn(3,4)
print(a)
```

```python
print('최댓값: ', torch.max(a))

# 옵션 dim : 각 차원에서 가장 큰 값 반환
print(torch.max(a, dim=0)) # 각 열에서 가장 큰 값들로만 구성된 행을 values의 1차원 텐서로 반환
print(torch.max(a, dim=1)) # 각 행에서 가장 큰 값들로만 구성된 열을 values의 1차원 텐서로 반환
# indices : 인덱스(dim=0 에서는 열 값, dim=1 에서는 행 값)
print()
# 옵션 keepdim : 차원을 그대로 유지하면서 최대/최소 반환(default: False)
print(torch.max(a, dim=0, keepdim= True))
print(torch.max(a, dim=1, keepdim= True))
print()

print(torch.min(a))
print(torch.min(a, dim=0, keepdim= True))
print(torch.min(a, dim=1, keepdim= True))

# arg = tensor(Index)
print(torch.argmax(a)) # tensor(5) 0행0열을 0으로 시작하여 1차원으로 펼쳤을 때 5번째 값
print(torch.argmin(a))
```

### 4. 통계

| 함수 | 의미 | 설명 |
| :---: | :---: | :---: |
| torch.mean(a) | 평균 | 전체원소의 평균 |
| torch.sum(a) | 합 | 전체원소의 합 |
| torch.prod(a) | 곱 | 전체원소의 곱 |
| torch.median(a) | 중앙값 | 정렬했을 때 가운데 값 |
| torch.std(a) | 표준편차 | 데이터의 분산 정도 |
| torch.var(a) | 분산 | 데이터의 분산 |

```python
print(a)

print('합계:', torch.sum(a))
print(torch.sum(a, dim=1))  # 각 행을 읽어와서 합계 계산
print(torch.sum(a, dim=1, keepdim=True)) # 차원을 유지하면서 각 행별 합계 계산
print()

print('평균:', torch.mean(a))
print(torch.mean(a, dim=1, keepdim=True))

print('곱:', torch.prod(a))

# 데이터의 갯수가 짝수이면 더 작은 값을 반환
print('중앙값:', torch.mean(a))

print('표준편차:', torch.std(a))

print('분산:', torch.var(a))
```

### 5. 삼각함수

| 함수 | 의미 | 설명 |
| :---: | :---: | :---: |
| torch.sin(a) | 사인 | sin(x) |
| torch.cos(a) | 코사인 | cos(x) |
| torch.tan(a) | 탄젠트 | tan(x) |
| torch.asin(a) | 아크사인 | sin의 역함수 |
| torch.acos(a) | 아크코사인 | cos의 역함수 |
| torch.atan(a) | 아크탄젠트 | tan의 역함수 |
| torch.tanh | 하이퍼볼릭 탄젠트 | 활성함수 |

### 하이퍼볼릭 탄젠트 정의 및 수식

* **수식**:
  $$\tanh(x) = \frac{\sinh(x)}{\cosh(x)} = \frac{e^{x}-e^{-x}}{e^{x}+e^{-x}}$$
* **시그모이드 함수와의 관계**:
  $$\tanh(x) = 2\text{sigmoid}(2x)-1$$

### 주요 특징

* **출력값의 범위**: $-1 \le \tanh(x) \le 1$ (시그모이드 함수의 범위인 $0 \sim 1$보다 넓음)
* **중심값(Center)**: $0$을 중심으로 대칭인 기함수(Odd function)입니다. (시그모이드는 $0.5$ 중심)
<br> : 중심값(0)에서 먼 값(10 등)을 대입하면 1 또는 -1에 가까운 값이 나오며
<br> : 중심값(0)에서 가까운 값(0.3 등)을 대입하면 원 값(0.3 등)에 거의 동일하게 출력
* **점근선**: $y = 1$ 및 $y = -1$

```python
print(torch.pi)
print(torch.sin(torch.tensor(torch.pi/6)))
print(torch.cos(torch.tensor(torch.pi/3)))
print(torch.tan(torch.tensor(torch.pi/4)))

print(torch.tanh(torch.tensor(-10)))

a = torch.tensor([-2.0, -1, -0.3, 0, 0.6, 1, 2])
print(torch.tanh(a))
```

### > torch.nan
- not a number(숫자가 아님) : 수학적으로 정의할 수 있으나 컴퓨터가 정상적인 실수값으로 표현할 수 없는 연산 불능 / 수치적인 오류 상태를 의미
    - 정의되지 않은 수학
        - 연산음수의 로그 계산 : torch.log(torch.tensor(-1.0)) -> nan
        - 0을 0으로 나누기 : torch.tensor(0) / torch.tensor(0) -> nan
        - 무한대 간의 뺄셈 : inf - inf -> nan

    - 딥러닝 학습 중 수치적 불안정성(Gradient Explosion 등)
        - 그래디언트 폭주 : 가중치 업데이트할 때마다 업데이트 양이 폭증하여 표현 범위 초과
        - 언더플로우 / 오버플로우

```python
torch.nan

print(torch.log(torch.tensor(-1)))

# isnan
print(torch.isnan(torch.tensor([1,2,torch.nan,4,5])))
# print(torch.isnan(torch.tensor([0/0]))) # zerodivisionerror는 tensor(nan) 아님
# isinf
print(torch.isinf(torch.tensor([1,2,3,4,torch.inf])))
```

#### > 기타 함수
### Transpose() : 2개의 차원 교환
 - 주로 2차원 행렬의 전치(Row <-> Column)나, 특정 두 축의 위치만 바꿀 때 사용
 - 차원의 개수와 상관없이 정확히 2개 축의 인덱스만 넘겨주어야 됨

 ### permute() : 모든 차원의 순서를 재배치
 - 3차원 이상의 고차원 텐서의 차원 순서를 한번에 뒤섞을 때 사용

 ### t(), T : 행렬의 연산(전치)를 간결하게 쓰기위한 메서드 및 속성
 - ~.t() : 2차원 텐서 전용 전치함수
 - .T : 모든 차원을 역순으로 뒤집은 결과를 반환
    - 2차원 행렬 : 일반적인 전치 행렬(A^T)이 됨
    - 고차원 텐서 : 차원의 순서가 완전히 반대로 뒤집힘(예) (2,3,4) -> (4,3,2)


```python
a = torch.tensor([1,2,3]) # 1차원
b = torch.tensor([2,2,1])
print(a)
print(b)
print(torch.sum(a*b))

a = a.reshape(3,1) # 2차원
b = b.reshape(3,1)
print(a)
print(b)
#transpose(0,1) : default -> 전치하려면 row와 colum의 axis를 바꾸면 됨
print(a.transpose(1,0)@b) # 1 x 3 @ 3 x 1 = 1 x 1, 여전히 2차원

print(a.permute(1,0))
print(a.T)
print(a.t())
```

```python
c = torch.randn(4,3,6)
print(c) # 각 원소들의 위치(x,y,z)
print(c.shape)

print(c.permute(1,2,0)) # 각원소들의 위치(y,z,x)
print(c.permute(1,2,0).shape)
```

### squeeze()
 - 크기가 1인 축을 제거하여 텐서의 차원을 축소할 때 사용
 - 데이터의 실제 원소 수나 값은 바꾸지 않고, 단지 불필요하게 감싸져있는 차원(1)만 골라서 차원의 수를 줄여주는 역할

 ### unsqueeze()
 - 지정한 위치에 크기가 1인 차원을 새롭게 추가하여 텐서의 차원을 확장할 때 사용
 - 실제 데이터나 원소의 개수는 그대로 유지하여 단지 차원의 모양만 변경

```python
a = torch.randn(1,1,1,3,1,1,1,4,1,1,1)
print(a.shape)
print(a.squeeze()) # 크기가 3,4인 차원만 남김
print(a.squeeze(dim=(0,1,2,3,4,5)).shape) # dim=(인덱스)의 값 중 크기가 1인 차원 삭제
print(a.squeeze(dim=(0,1,2,4,5,6,8,9,10)).shape)
```

```python
a = torch.randn(3,4)
print(a)
print(a.shape)

print(a.unsqueeze(dim=0).shape)
print(a.reshape(1,3,4).shape) # 결과 동일

print(a.unsqueeze(dim=1).shape)
print(a.reshape(3,1,4).shape)

print(a.unsqueeze(dim=2).shape)
print(a.reshape(3,4,1).shape)
```
