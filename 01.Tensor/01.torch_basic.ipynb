### PyTorch
 : 인공지능과 딥러닝 모델을 학습시키기 위한 파이썬 기반의 오픈소스 머신러닝 프레임워크

 - 핵심기능
    - Tensor : 데이터를 다차원 배열로 저장하고 계산
    - GPU 연산: GPU를 이용한 대량의 계산을 빠르게 처리( 구글  Colab에서 사용 )
    - 지동미분(autograd) : 학습에 필요한 기울기를 자동계산
    - 최적화(Optimizer) : 모델의 가중치를 수정하며 학습( 모델의 오차를 계산하는 손실함수를 최소화 )

### ▶ 초기화되지 않은 Tensor 생성

```python
x = torch.empty(4, 2)
x
```

### ▶ 0 또는 1로 채워진 Tensor

```python
x = torch.zeros(4, 2)
print(x)
print(x.dtype)

y = torch.ones(2, 4)
print(y)
print(y.dtype)
```

### ▶ 사용자가 입력한 값으로 Tensor 초기화

```python
a = torch.tensor([1,2,3,4]) # integer tensor 는 미분 불가
print(a)
print(type(a))
print(a.dtype)
print(a.shape)

# 하나라도 실수형이면 tensor 전체가 실수형이 된다.
b = torch.tensor([1,2,3.1,4]) # float tensor 는 미분 가능
print(b)
print(b.dtype)
```

### ▶ 기초함수

```python
c= torch.tensor(
    [[1,2,3],
     [4,5,6]]
)

# d = torch.tensor(
#     [[1,2], # 행렬이기에 각 행 열의 갯수가 같아야 한다.
#     [3,4,5]]
# )
# d # ValueError: expected sequence of length 2 at dim 1 (got 3)

print(c)
print(c.shape)
print(c.ndim) # 차원의 갯수는 깊이 = 대괄호 중첩의 갯수
print(c.numel()) # 총 원소의 갯수
# number of elements in the input tensor. This is equivalent to the product of the input tensor's
```

```python
print(torch.zeros(5))
print(f'Number of dimensions: {torch.zeros(5).ndim}')
# zeros_like() : 기존 tensor와 같은 shape, dtype, device를 가지는 tensor를 생성
print(torch.zeros_like(c))
print(f'Number of dimensions: {torch.zeros_like(c).ndim}')

print(torch.arange(3,10,2)) # range 와 동일한 tensor 생성
print(torch.arange(0,1,0.1)) # 0.0 ~ 1.0 까지 0.1 간격으로 tensor 생성

print(torch.linspace(0,1,10)) # 0.0 ~ 1.0 까지 10개의 tensor 생성 (end 값 포함)
```

### ▶ Tensor의 연산
: tensor에 대한 수학연산 삼각함수 비트연산 비교연산 집계 등 제공

```python
#  1차원 연산
d = torch.tensor([1,3,3])
e = torch.tensor([4,5,6])
f = d + e
f
```

```python
# 2차원 연산
g = torch.tensor([[1,2,3],
                  [1,2,3]])
h = torch.tensor([[4,5,6],
                  [1,1,1]])
i = g + h
print(i)

j = g - h
print(j)

print(g*h) # 행렬곱 아님에 주의, 각 element 별 곱셈
print(g/h)
print(g**2)
```

### ▶ 행렬곱(dot product) 
   : torch.mm() or torch.matmul() or @

```python
k = torch.tensor([[1,2],
                  [3,4]])
m = torch.tensor([[1,2],
                  [3,4]])
print (k @ m) # 행렬곱
print (k.matmul(m)) # 행렬곱
print (torch.mm(k,m)) # 행렬곱
```
