#### > DataLoader

```python
import torch
import numpy as np
import pandas as pd
from torch.utils.data import DataLoader, TensorDataset
```

#### (1) 데이터 준비

```python
path = 'https://raw.githubusercontent.com/DA4BAM/dataset/master/advertising.csv'
data = pd.read_csv(path)
print(data.head())

data_input = data[:10] # data에서 10줄만 가져와서 저장

# x, y 분할
target = 'Sales'
x = data_input.drop(target, axis=1)
print(x)
y = data_input.loc[:, target] # Sales의 데이터만 추출
print(y)
print(x.shape, y.shape)
```

#### (2) tensor로 변경

```python
print(type(x)) # DataFrame
print(type(y)) # Series

# numpy 로 변환 (Scaling 단계지만 생략)
# 딥러닝에서는 텐서 변환을 위해 데이터 세트가 반드시 numpy배열이어야 함
x_np = x.values
y_np = y.values

print(type(x_np), type(y_np))
print(x_np.dtype, y_np.dtype)
print(x_np.shape, y_np.shape)

y_np = y_np.reshape(-1,1)
print(y_np.shape)
```

```python
# Tensor로 변경
# dtype : numpy(float 64) -> torch는 float32를 기본적인 가중치로 사용
x_tensor = torch.tensor(x_np, dtype=torch.float32)
y_tensor = torch.tensor(y_np, dtype=torch.float32)

print(x_tensor.shape, y_tensor.shape)
```

### (3) DataLoader 준비
 - x_tensor, y_tensor를 하나의 tensor dataset으로 합치기
 - DataLoader 생성

1) Tensor DataSet

```python
dataset = TensorDataset(x_tensor, y_tensor)

print(dataset[:5]) # 5행 조회

print(dataset.tensors) # 전체 조회

print(dataset.tensors[0]) # x_tensor 조회
print(dataset.tensors[1]) # y_tensor 조회

print(len(dataset)) # 데이터 건수 조회
```

```python
data_loader = DataLoader(dataset, batch_size=3, shuffle=True)
```

batch_size = 3 : 한번에 모델에 전달할 데이터의 개수를 지정<br>
shuffle = : 학습데이터를 매 주기(epoch)시작시 무작위로 섞을지 지정<br>
&emsp;&emsp; True : 데이터를 무작위로 섞음<br>
&emsp;&emsp; False : 데이터의 원래 순서를 유지<br><br>

에포크(epoch): 주기, 전체 학습데이터를 모델이 한번 모두 학습하는 것<br>
&emsp; 예를 들어 데이터가 8개이고 batch_size=3 이라면,<br>
&emsp;&emsp; Batch 1 : 데이터 3개 학습<br>
&emsp;&emsp; Batch 2 : 데이터 3개 학습<br>
&emsp;&emsp; Batch 3 : 데이터 2개 학습<br>
&emsp; 위 학습 과정이 끝나면 1 epoch 완료

```python
# 데이터 로더로부터 데이터 추출
for x_b, y_b in data_loader:
    print(x_b)
    print(y_b)
    break 
# batch_size=3 이므로 3개 세트의 데이터가 입력됨
# 순서는 무작위 (실행할때마다 값 변함)
```

```python
# 데이터 로더로부터 배치 단위로 데이터 로딩
for batch_id, (x_b, y_b) in enumerate(data_loader): # enumerate : index번호를 key로 부여
    print(f'batch_id: {batch_id}')
    print(x_b)
    print(y_b)
    print('-'*50)
```

```python
print(f'데이터로더의 데이터 세트 수: {len(data_loader.dataset)}')
print(f'데이터로더의 배치 수: {len(data_loader)}')
```
