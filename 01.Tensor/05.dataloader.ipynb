#### > DataLoader

```python
import torch
import numpy as np
import pandas as pd
from torch.utils.data import DataLoader, TensorDataset
```

#### (1) 데이터 준비

```python
path = 'https://raw.githubusercontent.com/DA4BAM/dataset/master/advertising.csv'
data = pd.read_csv(path)
print(data.head())

data_input = data[:10] # data에서 10줄만 가져와서 저장

# x, y 분할
target = 'Sales'
x = data_input.drop(target, axis=1)
print(x)
y = data_input.loc[:, target] # Sales의 데이터만 추출
print(y)
print(x.shape, y.shape)
```

#### (2) tensor로 변경

```python
print(type(x)) # DataFrame
print(type(y)) # Series

# numpy 로 변환 (딥러닝에서는 텐서 변환을 위해 데이터 세트가 반드시 numpy배열이어야 함)
x_np = x.values
y_np = y.values

print(type(x_np), type(y_np))
print(x_np.dtype, y_np.dtype)
print(x_np.shape, y_np.shape)

y_np = y_np.reshape(-1,1)
print(y_np.shape)
```
