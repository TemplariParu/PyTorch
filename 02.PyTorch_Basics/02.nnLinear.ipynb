#### > 인공 신경망
- nn 모듈 : neural network (신경망) 을 구축하고 학습시키는데 필요한 모든 구성요소를 제공하는 핵심 모듈
- 주요 역할 및 구성 요소
    1. 인공 신경망 층(layer) : 입력데이터를 변환하는 여러개의 층
        - 선형 layer ( nn.Linear ) : 완전연결층 (full connected layer)
        - 합성곱 layer ( nn.Con2d ) : 이미지 처리에 쓰이는 CNN 레이어
        - 순환 layer ( nn.LSTM, nn.GRU ) : 텍스트, 시계열 처리에 쓰이는 RNN 레이어

    2. 활성화 함수(Activation Functions) : 비선형성을 추가하여 모델이 복잡한 패턴을 학습할 수 있도록 해준다.
        - nn.ReLU(), nn.Sigmoid(), nn.Softmax() 등등
    3. 손실 함수(Loss Functions) : 모델의 예측값과 실제값의 오차를 계산
        - nn.MESloss(), nn.CrossEntropyloss()
    4. 컨테이너 : 여러 레이어를 묶어서 하나의 모델로 구조화
        - nn.Module : 모든 신경망 모델의 기본 클래스
            - 자신이 만드는 모든 커스텀 모델은 nn.Module을 상속받아 정의하여 사용
        - nn.Sequential : 레이어를 순차적으로 연결해주는 간단한 래퍼


```python
import torch
from torch import nn
from torchviz import make_dot
```

```python
x = torch.tensor([1.])

model = nn.Linear(1, 1) # 입력값(in_features) 1개, 출력값(out_features) 1개 인 layer 생성
print(model)

print('w= ', model.weight)
print('b= ', model.bias)
print()

y = model(x)
print(y) # y = wx + b | y = x @ model.weight + model.bias

y2 = x @ model.weight + model.bias
print('y2 :', y2)
```
