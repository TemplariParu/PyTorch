### > PyTorch의 함수들

### > 난수 생성 함수

| 함수 | 분포 | 범위 및 특징 |
| :---: | :---: | :---: |
| torch.rand() | 균등분포 | 0~1 |
| torch.randn() | 정규분포 | 평균0, sd 1 |
| torch.randint() | 이산균등분포 | 정수 난수|

```python
import torch
```

```python
print(torch.rand(3,3))
print(torch.randn(3,3))
print(torch.randint(3,10,(5,)))
print(torch.randint(3,10,(2,3)))
```

#### > PyTorch 주요 수학 함수

### 1. 기본 수학 함수

| 함수 | 의미 | 예시 | 설명 |
| :---: | :---: | :---: | :---: |
| torch.abs(a) | 절대값 | abx(-3) -> 3 | 모든 원소의 절대값 반환 |
| torch.sqrt(a) | 제곱근 | sqrt(9) -> 3 | $\sqrt(x)$ |
| torch.square(a) | 제곱 | square(3) -> 9| $x^2$ |
| torch.pow(a, n) | 거듭제곱 | pow(2,3) -> 8 | $x^n$ |
| torch.exp(a) | 지수함수 | exp(1) -> e | $e^x$ |
| torch.log(a) | 자연로그 | log(e) -> 1 | ln(x) |
| torch.log10(a) | 상용로그 | log10(10) -> 1 | $log_{10}(x)$ |
| torch.log2(a) | 밑이 2인로그 | log2(2) -> 1 | $log_{2}(x)$ |

```python
a = torch.randn(3,3)
print(a)

# 기본 수학함수
print(torch.abs(a))
print(torch.sqrt(torch.abs(a)))
print(torch.square(a))
print(torch.pow(a, 2))
print()
# 지수함수
print(torch.exp(a))
print(torch.log(torch.abs(a))) # logₑa
print(torch.log(torch.exp(torch.tensor(1))))
print(torch.log(torch.exp(torch.tensor(a)))) # logₑeᵃ
print(torch.log10(a))
print(torch.log2(a))
```

### 2. 반올림 / 올림 / 내림

| 함수 | 의미 | 예시 |
| :---: | :---: | :---: |
| torch.round(a) | 반올림 | round(1.6) -> 2 |
| torch.round(a, decimals=2) | 지정한 소수점까지 반올림 | round(1.6275, 2) -> 1.63 |
| torch.floor(a) | 내림 | floor(1.9) -> 1 |
| torch.ceil(a) | 올림 | ceil(1.1) -> 2 |
| torch.trunc(a) | 소수점버림 | trunc(1.87) -> 1 |

```python
print(torch.round(a))

print(torch.round(a, decimals=2))
print(torch.floor(a))
print(torch.ceil(a))
print(torch.trunc(a))
```

### 3. 최솟값 / 최댓값

| 함수 | 의미 | 설명 |
| :---: | :---: | :---: |
| torch.min(a) | 최솟값 | 전체원소 중 가장 작은 값 |
| torch.max(a) | 최댓값 | 전체원소 중 가장 큰 값 |
| torch.argmin(a) | 최솟값의 인덱스 | 최솟값이 위치한 인덱스 반환 |
| torch.argmax(a) | 최댓값의 인덱스 | 최댓값이 위치한 인덱스 반환 |

```python
a = torch.randn(3,4)
print(a)
```

```python
print('최댓값: ', torch.max(a))

# 옵션 dim : 각 차원에서 가장 큰 값 반환
print(torch.max(a, dim=0)) # 각 열에서 가장 큰 값들로만 구성된 행을 values의 1차원 텐서로 반환
print(torch.max(a, dim=1)) # 각 행에서 가장 큰 값들로만 구성된 열을 values의 1차원 텐서로 반환
# indices : 인덱스(dim=0 에서는 열 값, dim=1 에서는 행 값)
print()
# 옵션 keepdim : 차원을 그대로 유지하면서 최대/최소 반환(default: False)
print(torch.max(a, dim=0, keepdim= True))
print(torch.max(a, dim=1, keepdim= True))
print()

print(torch.min(a))
print(torch.min(a, dim=0, keepdim= True))
print(torch.min(a, dim=1, keepdim= True))

# arg = tensor(Index)
print(torch.argmax(a)) # tensor(5) 0행0열을 0으로 시작하여 1차원으로 펼쳤을 때 5번째 값
print(torch.argmin(a))
```
