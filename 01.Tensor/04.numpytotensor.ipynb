### > Numpy 배열을 Tensor로 변환
: 임베딩 벡터 처리, 데이터 전처리, 모델 입력 데이터 구성 등에서 많이 사용

```python
import numpy as np
import torch
```

- torch.from_numpy(array): 메모리 공유, numpy 배열을 tensor로 변환

```python
arr = np.array([1.0, 2,3,4])
x = torch.from_numpy(arr)

print('numpy 배열:', arr)
print('tensor:', x)

arr[0] = 100
print('변경후')
print('numpy 배열:', arr)
print('tensor:', x)
```

- torch.tensor(array)
    : 데이터를 복사, 원본과 분리된 tensor 생성

```python
arr = np.array([1.0, 2,3,4])
x = torch.tensor(arr)

print('numpy 배열:', arr)
print('tensor:', x)

arr[0] = 100
print('변경후')
print('numpy 배열:', arr)
print('tensor:', x) # tensor의 값은 arr의 변경과 무관함을 알 수 있음
```

- torch.as_tensor(array)
    : 가능하면 메모리를 공유하고, 필요한 경우에만 데이터를 복사하는 함수
    - torch.from_numpy()와 비슷하게 동작하지만, 입력자료형이나, 장치(device)등을 지정할 수 있다는 차이

```python
arr = np.array([1.0, 2,3,4])
x = torch.as_tensor(arr)

print('numpy 배열:', arr)
print('tensor:', x)

arr[0] = 100
print('변경후')
print('numpy 배열:', arr)
print('tensor:', x) # from_numpy처럼 메모리 공유
```

```python
arr = np.array([1,2,3,4], dtype=np.int32)

x = torch.as_tensor(arr, dtype= torch.float32)

print('numpy 배열:', arr)
print('tensor:', x)

arr[0] = 100
print('변경후')
print('numpy 배열:', arr)
print('tensor:', x) # 타입 변경하면 새로운 텐서로 저장
```

#### > PyTorch Tensor를 Numpy 배열로 변환
: 모델이 계산한 결과를 python의 다른 library로 전달을 하거나, 데이터를 시각화하고 분석할 때 사용


```python
x = torch.tensor([.1,.2,.3,.4])
arr = x.numpy()

print('numpy 배열:', arr)
print('tensor:', x)
print('자료형:', arr.dtype)
print('크기:', arr.shape)

x[0] = 0.7
print('변경후')
print('numpy 배열:', arr)   # 마찬가지로 변경됨(메모리 공유)
print('tensor:', x)
print('자료형:', arr.dtype) 
print('크기:', arr.shape)
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

x = torch.tensor([1.,2,3], device= device)

# .cpu(): GPU에 있던 데이터를 CPU로 옮기는 코드
# arr = x.cpu().numpy()
```
